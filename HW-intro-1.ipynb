```python
import numpy as np
import pandas as pd

# 1. Select all the cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# 2. Select only columns vehicle_weight and model_year
selected_df = asia_cars[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
first_7 = selected_df.head(7)

# 4. Get the underlying NumPy array. Let's call it X.
X = first_7.values

# 5. Compute matrix-matrix multiplication between the transpose of X and X (XTX)
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply the inverse of XTX with the transpose of X, and then multiply by y (w)
w = (XTX_inv @ X.T) @ y

# 9. What's the sum of all the elements of the result?
w_sum = w.sum()

print("Weights vector (w):", w)
print("Sum of weights:", w_sum)
```
